```python
import pandas as pd
```

```python
df=pd.read_csv(r"D:\Machine_Learning_Journey\EDA\Employee\employee_data.csv")
df['gender']=['Male','Female','Male','Female','Male','Female','Male','Female','Male','Female']
```

```python
#Feature engineering 
#1. Apply 
df['tax']=df['Salary'].apply(lambda x:"20%" if x>50000 else "10%")
df
```

```python
# 2. Map 
gender={'Male':"M","Female":"F"}
df["gender"]=df['gender'].map(gender)
df
```

```python
# 3.Assign 
df.assign(new_income=df['Salary']*10)
```

```python
# 4.Replace
df['Department'].replace("HR",'hr')
```

```python
df['Salary']=df['Salary'].sort_values( )
df['Ranking']=df['Salary'].rank(ascending=False,method='dense')
df
```

```python

df['Salary']=df['Salary'].sort_values( )
df
```
