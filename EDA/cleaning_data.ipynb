```python
import json
```

```python
def loading_data(source):
    with open(source,'r') as f:
        data=json.load(f)
    return data
data=loading_data("D:\Machine_Learning_Journey\EDA\customer_data.json")
```

```python
def clean_data(data):
    text_to_num = {"one" : 1, "two" : 2, "three" : 3, "four": 4, "five": 5}
    for user in data:
        raw_rating=str(user['rating']).strip().lower()
        if(raw_rating in text_to_num):
            user['rating']=text_to_num[raw_rating]
    return data
data=clean_data(data)
    
```

```python
#remove  duplicate entries
def remove_duplicate(source):
    unique=set()
    clean_data=[]
    for user in source:
        if user['name'].strip() in unique:
            continue
        unique.add(user['name'])
        clean_data.append(user)
    return clean_data
    
```

```python
data=remove_duplicate(data)
path = r"D:\Machine_Learning_Journey\EDA\customer_data.json"

with open(path, "w") as file:
    json.dump(data, file, indent=4)
```

```python
def average_rating(source):
    sum_of_rating=0
    for user  in source:
        sum_of_rating += float(user['rating'])
    return sum_of_rating/len(source)
average_rating(data)
```

```python
#percentage of users with poor rating (<3)
def poor_rating_users(data):
    user_count = 0
    for user in data:
        if float(user['rating'])<3:
            user_count += 1
    return (user_count/len(data))*100
poor_rating_users(data)
```

```python
#Building recommendation system
#if user rating>=4 recommend same brand product
# else recommend different brand product
def recommendation(data):
    recommendated=[]
    for user in data:
        curr_recommend={}
        curr_recommend['user']=user['name']
        if float(user['rating'])>=4:
            curr_recommend['brand']='Apple'
        else:
            curr_recommend['brand']='samsung'
        recommendated.append(curr_recommend)
    return recommendated
recommendation(data)
```
