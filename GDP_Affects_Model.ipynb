# Predicting how GDP affects happiness using linear regression

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LinearRegression
#  Load the dataset
data_url = "https://github.com/ageron/data/raw/main/"
life = pd.read_csv(data_url + "lifesat/lifesat.csv")
# X = GDP per capita (feature/input)
# Y = Life satisfaction (target/output)
X = life[["GDP per capita (USD)"]].values
Y = life[["Life satisfaction"]].values
life.plot(kind="scatter", grid=True,
          x="GDP per capita (USD)",
          y="Life satisfaction")
plt.axis([23500, 62500, 4, 9])
model = LinearRegression()
model.fit(X, Y)
#  Print model parameters (to understand learning)
print("Slope (m):", model.coef_)
print("Intercept (b):", model.intercept_)
#  Plot the prediction line
X_range = np.linspace(23500, 62500, 100).reshape(-1, 1)
Y_pred = model.predict(X_range)
plt.plot(X_range, Y_pred, color="red")
# Step 7: Predict for new GDP value  
X_new = [[37655.2]]
prediction = model.predict(X_new)
print("\nPrediction for Cyprus GDP:")
print("GDP:", X_new[0][0])
print("Predicted Life Satisfaction:", prediction[0][0])
# Try multiple predictions  
test_values = [[20000], [40000], [60000]]
print("\nSome more predictions:")
for val in test_values:
    pred = model.predict([val])
    print(f"GDP: {val[0]} → Predicted Happiness: {pred[0][0]}")
plt.show()
```
